```python
dbutils.widgets.text("catalog", "fk_1361")
catalog = dbutils.widgets.get("catalog")
```

### Load Stores data 

```python
spark.sql(f"""
create table if not exists {catalog}.bronze.stores_bronze
""")
```

```python
spark.sql(f"""
copy into {catalog}.bronze.stores_bronze
from (select *, _metadata.file_path as file_path, _metadata.file_name as file_name, current_timestamp() as ingestion_ts from '/Volumes/{catalog}/landing/raw/stores/')
fileformat = csv
format_options ('header' = 'true')
copy_options ('mergeSchema' = 'true')
""")
```

### Load Customers data
